# Validating Generated Code

Validate one code file directly, then run batch validation against a test group using local runtime.

```python
from pathlib import Path

from faas_gauge.config import get_data_dir
from faas_gauge.experiment import validate_batch
from faas_gauge.experiment.validate import validate_code
from faas_gauge.notebooks.helpers import print_validation_summary

data_dir = get_data_dir()
```

```python
sample_code = Path("/tmp/faas_gauge_notebook_sample.py")
sample_code.write_text("def solve(n):\n    return n * 2\n", encoding="utf-8")

single_result = validate_code(
    code_path=str(sample_code),
    runtime="local",
    iterations=1,
    max_runtime=10,
    test_data={"tests": [{"name": "double", "input": {"n": 3}, "expected": 6}]},
)

single_result["summary"]
```

```python
batch_result = validate_batch(
    test_group="notebook-demo",
    iterations=1,
    runtime="local",
    max_runtime=30,
    data_dir=data_dir,
)

print_validation_summary(batch_result["batch_id"], batch_result["summary"])
```

```python
print("batch id:", batch_result["batch_id"])
print("total tests:", batch_result["summary"]["total_tests"])
print("passed:", batch_result["summary"]["passed"])
print("failed:", batch_result["summary"]["failed"])
print("errors:", batch_result["summary"]["errors"])
```
