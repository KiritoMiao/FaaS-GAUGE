# Revolution Quickstart

Get running in about five minutes: import the package, load credentials, inspect providers, and run one AI request.

```python
import faas_gauge

print("faas_gauge version:", faas_gauge.__version__)
```

```python
from pathlib import Path

from faas_gauge.config import get_data_dir, load_credentials

data_dir = get_data_dir()
creds = load_credentials()

print("data dir:", data_dir)
print("credential providers:", sorted(creds.keys()))
```

```python
from faas_gauge.ai import AIClient

client = AIClient()
providers = client.providers()
print("providers:", providers)

if providers:
    first_provider = providers[0]
    print(f"models for {first_provider}:")
    print(client.models(first_provider))
```

```python
question = "Write a Python function add(a, b) that returns a + b."
provider = providers[0]
model = client.models(provider)[0]

response = client.ask(question=question, provider=provider, model=model)

print("provider/model:", response.provider, response.model)
print("error:", response.error)
print("tokens in/out:", response.input_tokens, response.output_tokens)
print("rtt sec:", round(response.rtt_sec, 2))
print("python code preview:")
print(response.python_code[:500])
```
