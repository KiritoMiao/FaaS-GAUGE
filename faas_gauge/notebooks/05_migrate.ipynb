# Migrating from Legacy System

Migrate legacy SQLite and YAML credentials into the JSON/JSONL data layout used by faas_gauge.

Legacy inputs commonly include:

- SQLite DB with `experiments`, `iteration_results`, and validation tables
- `.secret.yaml` for credentials
- old `question/` and `test_data/` directories

````python
from pathlib import Path

from faas_gauge.config import get_data_dir
from faas_gauge.store import ExperimentStore
from faas_gauge.store.migration import (
    migrate_all,
    migrate_credentials,
    migrate_from_sqlite,
)

target_data_dir = get_data_dir()
legacy_db = Path("/path/to/legacy.db")
legacy_root = Path("/path/to/legacy_project")
secret_yaml = legacy_root / ".secret.yaml"
````

````python
store = ExperimentStore(data_dir=target_data_dir)
experiments = store.list_experiments()
validations = store.list_validations()

print("migrated experiments:", len(experiments))
print("migrated validation batches:", len(validations))
print("credentials path:", target_data_dir / "config" / "credentials.json")
````

### Script usage

You can also run migration from CLI:

```bash
python scripts/migrate.py --db /path/to/legacy.db --source-dir /path/to/legacy_project -t data
python scripts/migrate.py --db /path/to/legacy.db --source-dir /path/to/legacy_project -t data --only experiments
python scripts/migrate.py --db /path/to/legacy.db --source-dir /path/to/legacy_project -t data --only credentials -s /path/to/.secret.yaml
```
