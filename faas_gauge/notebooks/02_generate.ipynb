# Generating Code with AI

Run code generation workflows and inspect experiment metadata plus iteration-level metrics.

```python
from pathlib import Path

from faas_gauge.config import get_data_dir
from faas_gauge.experiment import generate, load_experiment
from faas_gauge.notebooks.helpers import format_iteration_table, print_experiment_summary

data_dir = get_data_dir()
question_path = data_dir / "questions" / "prime_number_generator.txt"
question_content = question_path.read_text(encoding="utf-8")

print("using question file:", question_path)
```

```python
single = generate(
    question="prime_number_generator",
    question_content=question_content,
    provider="openai",
    model="gpt-5-mini",
    iterations=3,
    max_workers=1,
    test_group="notebook-demo",
    data_dir=data_dir,
)

print_experiment_summary(single["id"], single["summary"])
```

```python
exp = load_experiment(single["id"], data_dir=data_dir)
iteration_rows = list(exp.iterations())
print(format_iteration_table(iteration_rows, max_rows=10))
```

```python
parallel = generate(
    question="prime_number_generator",
    question_content=question_content,
    provider="openai",
    model="gpt-5-mini",
    iterations=8,
    max_workers=4,
    test_group="notebook-demo",
    data_dir=data_dir,
)

print_experiment_summary(parallel["id"], parallel["summary"])
```

### Notes

- `max_workers=1` runs single-threaded.
- `max_workers>1` uses a thread pool and writes all iteration records to JSONL.
